```python
import os
import sys
import pandas as pd

# 1. Tự động dò tìm thư mục gốc của project (nơi có chứa thư mục 'src')
PROJECT_ROOT = os.getcwd()
while not os.path.exists(os.path.join(PROJECT_ROOT, 'src')):
    PROJECT_ROOT = os.path.dirname(PROJECT_ROOT)
    # Tránh loop vô hạn nếu lỡ để project ngoài desktop
    if PROJECT_ROOT == os.path.dirname(PROJECT_ROOT):
        break

# 2. Set thư mục làm việc và đưa vào sys.path
if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)
os.chdir(PROJECT_ROOT)

print(f"✅ Đã set thư mục làm việc tại: {os.getcwd()}")

# 3. Import hàm từ thư mục src
from src.models.supervised import train_and_evaluate

# 4. Đọc dữ liệu
try:
    df_ml = pd.read_csv('data/processed/hr_ml_encoded.csv') 
    print("✅ Đã load dữ liệu thành công!")
except FileNotFoundError:
    print("🚨 Lỗi: Không tìm thấy file, bác kiểm tra lại đường dẫn nhé!")

# --- THÊM DÒNG NÀY ĐỂ CHỐNG RÒ RỈ DỮ LIỆU ---
if 'Attrition_No' in df_ml.columns:
    df_ml = df_ml.drop(columns=['Attrition_No'])

# 5. Chạy mô hình (SỬA 'Attrition' THÀNH 'Attrition_Yes')
results, X_test, y_test = train_and_evaluate(df_ml, target_col='Attrition_Yes', config_path='configs/params.yaml')

# 6. In báo cáo
if results:
    for name, metrics in results.items():
        print("\n" + "="*50)
        print(f"BÁO CÁO CHI TIẾT: {name}")
        print(metrics['report'])
```
