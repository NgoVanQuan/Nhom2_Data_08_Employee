# KHAI PHÁ DỮ LIỆU & PHÂN CỤM NHÂN SỰ (DATA MINING)

File này chịu trách nhiệm thực hiện 2 kỹ thuật học không giám sát cốt lõi:
1. **Khai phá luật kết hợp (Association Rules):** Sử dụng thuật toán FP-Growth để tìm ra các "tổ hợp nguyên nhân" dẫn đến việc nhân viên nghỉ việc (Leave) hoặc tiếp tục gắn bó (Stay).
2. **Phân cụm (Clustering):** Sử dụng K-Means và phương pháp Elbow để phân mảnh nhân sự thành các nhóm (Persona) có hành vi và đặc trưng tương đồng, từ đó xây dựng bảng Profiling nhận diện rủi ro.

```python
import sys
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from sklearn.cluster import KMeans

# Chỉ đường cho Python nhận diện thư mục src/
sys.path.append(os.path.abspath('..'))
from src.mining.miner import Miner

warnings.filterwarnings('ignore')
sns.set_theme(style="whitegrid") # Cài đặt style biểu đồ cho đẹp

print("✅ Đã nạp thành công các thư viện và class Miner!")
```

## BƯỚC 1: NẠP DỮ LIỆU ĐÃ TIỀN XỬ LÝ
Tải 2 tập dữ liệu đã được xử lý từ File 02:
* `hr_rules_binned.csv`: Dữ liệu dạng chữ (đã binning) dành cho FP-Growth.
* `hr_ml_encoded.csv`: Dữ liệu dạng số (One-hot encoding) dành cho K-Means.

```python
print("Đang nạp dữ liệu từ thư mục data/processed/...")

df_rules = pd.read_csv('data/processed/hr_rules_binned.csv')
df_ml = pd.read_csv('data/processed/hr_ml_encoded.csv')

print(f"-> Dữ liệu Luật kết hợp: {df_rules.shape}")
print(f"-> Dữ liệu Phân cụm & ML: {df_ml.shape}")

# Khởi tạo thuật toán từ src/ (Thiết lập tham số theo đúng báo cáo)
miner = Miner(min_support=0.15, max_len=3, n_clusters=3)
```

## BƯỚC 2: KHAI PHÁ LUẬT KẾT HỢP VỚI FP-GROWTH
Sử dụng độ đo Lift để đánh giá mức độ tương quan. Lọc ra Top 5 luật giá trị nhất cho 2 kịch bản: Nghỉ việc và Ở lại.

```python
# Chạy thuật toán FP-Growth
rules = miner.mine_association_rules(df_rules)
cols_to_show = ['antecedents', 'consequents', 'support', 'confidence', 'lift']

# --- TOP 5 LUẬT NGHỈ VIỆC (LEAVE) ---
print("🔴 TOP 5 TỔ HỢP NGUYÊN NHÂN DẪN ĐẾN NGHỈ VIỆC (LEAVE):")
leave_rules = rules[rules['consequents'].astype(str).str.contains('Attrition_Yes', na=False)]
display(leave_rules.sort_values(by='lift', ascending=False)[cols_to_show].head(5))

# --- TOP 5 LUẬT Ở LẠI (STAY) ---
print("\n🟢 TOP 5 TỔ HỢP NGUYÊN NHÂN GIỮ CHÂN NHÂN VIÊN (STAY):")
stay_rules = rules[rules['consequents'].astype(str).str.contains('Attrition_No', na=False)]
display(stay_rules.sort_values(by='lift', ascending=False)[cols_to_show].head(5))
```

## BƯỚC 3: XÁC ĐỊNH SỐ CỤM BẰNG PHƯƠNG PHÁP ELBOW
Tính toán Tổng bình phương khoảng cách (WCSS) với K chạy từ 2 đến 7 để tìm điểm "khuỷu tay".

```python
print("Đang tính toán phương pháp Elbow...")

# Lấy dữ liệu dạng số, bỏ biến mục tiêu để tính khoảng cách
features_elbow = df_ml.drop(columns=[c for c in df_ml.columns if 'Attrition' in c], errors='ignore')

# Chuẩn hóa tạm thời để vẽ Elbow
X_scaled_elbow = miner.scaler.fit_transform(features_elbow)

wcss = []
k_range = range(2, 8)
for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled_elbow)
    wcss.append(km.inertia_)

# Trực quan hóa Elbow
plt.figure(figsize=(9, 5))
plt.plot(k_range, wcss, marker='o', linestyle='--', color='blue', markersize=8)
plt.title('Phương pháp Elbow (Khuỷu tay) tìm số cụm K', fontsize=14)
plt.xlabel('Số lượng cụm (K)', fontsize=12)
plt.ylabel('Tổng bình phương khoảng cách (WCSS)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

print("-> KẾT LUẬN: Đồ thị gãy gập rõ nhất tại K=3. Chọn K=3 làm tham số chính thức.")
```

## BƯỚC 4: PHÂN CỤM K-MEANS & XÂY DỰNG BẢNG PROFILING
Tiến hành phân cụm với K=3 và tổng hợp giá trị trung bình các đặc trưng cốt lõi để phác họa chân dung từng cụm.

```python
# Chạy K-Means từ class Miner
labels = miner.run_kmeans(df_ml)

# Ghép nhãn Cluster vào dữ liệu để phân tích
df_profiling = df_rules.copy()
df_profiling['Cluster'] = labels

# Ép kiểu Attrition (Yes=1, No=0) để tính tỷ lệ nghỉ việc trung bình
df_profiling['Attrition_Rate'] = df_profiling['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)

print("\n📊 BẢNG PROFILING: CHÂN DUNG CÁC CỤM (Trung bình các đặc trưng):")
cols_to_profile = ['Age', 'MonthlyIncome', 'TotalWorkingYears', 'YearsAtCompany', 'DistanceFromHome', 'Attrition_Rate']
profiling_table = df_profiling.groupby('Cluster')[cols_to_profile].mean().round(2)

# Đổi tên cột Attrition_Rate lại thành Attrition cho giống file Word
profiling_table = profiling_table.rename(columns={'Attrition_Rate': 'Attrition'})
display(profiling_table)
```
