```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import warnings
warnings.filterwarnings('ignore')

from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import precision_recall_curve, average_precision_score

# 1. Tự động dò tìm thư mục gốc của project (nơi chứa thư mục 'src')
PROJECT_ROOT = os.getcwd()
while not os.path.exists(os.path.join(PROJECT_ROOT, 'src')):
    PROJECT_ROOT = os.path.dirname(PROJECT_ROOT)
    if PROJECT_ROOT == os.path.dirname(PROJECT_ROOT):
        break

# 2. Set thư mục làm việc về gốc dự án
if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)
os.chdir(PROJECT_ROOT)

print(f"✅ Đã set thư mục làm việc tại gốc dự án: {os.getcwd()}")

# Thiết lập phong cách vẽ biểu đồ
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
# Đường dẫn dữ liệu và mô hình
X_test_path = '../data/processed/X_test.csv'
y_test_path = '../data/processed/y_test.csv'
model_path = '../models/best_model_rf.pkl' # Đổi tên file mô hình nếu cần

try:
    X_test = pd.read_csv(X_test_path)
    y_test = pd.read_csv(y_test_path).values.ravel()
    
    model = joblib.load(model_path)
    print("✅ Đã tải dữ liệu Test và Mô hình thành công!")
except Exception as e:
    print(f"❌ Lỗi khi tải file: {e}")
```

```python
# Dự đoán nhãn và xác suất
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]

print("="*55)
print("BÁO CÁO ĐÁNH GIÁ MÔ HÌNH DỰ ĐOÁN NGHỈ VIỆC (SUPERVISED)")
print("="*55)
print(classification_report(y_test, y_pred, target_names=['Ở lại (0)', 'Nghỉ việc (1)']))
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Reds', cbar=False,
            xticklabels=['Dự đoán: Ở lại', 'Dự đoán: Nghỉ việc'],
            yticklabels=['Thực tế: Ở lại', 'Thực tế: Nghỉ việc'])
plt.title('Ma trận nhầm lẫn (Confusion Matrix) - Nhận diện nhân viên nghỉ việc', fontsize=14)
plt.yticks(rotation=0)
plt.show()
```

```python
precision, recall, _ = precision_recall_curve(y_test, y_proba)
pr_auc = average_precision_score(y_test, y_proba)

plt.figure(figsize=(8, 6))
plt.plot(recall, precision, color='purple', lw=2, label=f'PR Curve (AUC = {pr_auc:.3f})')
plt.xlabel('Recall (Tỷ lệ phát hiện đúng người nghỉ)')
plt.ylabel('Precision (Độ chính xác khi dự đoán nghỉ)')
plt.title('Đường cong Precision-Recall (PR-AUC)', fontsize=14)
plt.legend(loc="lower left")
plt.show()
```

```python
# Gắn kết quả dự đoán vào tập X_test ban đầu để phân tích
error_analysis_df = X_test.copy()
error_analysis_df['Thực_tế'] = y_test
error_analysis_df['Dự_đoán'] = y_pred

# Lọc ra những người THỰC TẾ NGHỈ nhưng MÔ HÌNH ĐOÁN Ở LẠI (False Negatives - Bỏ lọt rủi ro)
false_negatives = error_analysis_df[(error_analysis_df['Thực_tế'] == 1) & (error_analysis_df['Dự_đoán'] == 0)]

print(f"Số lượng nhân viên bị bỏ lọt (False Negatives): {len(false_negatives)} người.")
print("\nĐặc điểm trung bình của nhóm bị bỏ lọt này (Để tìm ra lý do mô hình đoán sai):")
# Hiển thị một vài cột quan trọng (Sửa lại tên cột nếu tên trong file của bạn khác)
cols_to_check = ['Age', 'MonthlyIncome', 'YearsAtCompany', 'DistanceFromHome']
# Chỉ lấy những cột có tồn tại trong X_test
cols_exist = [col for col in cols_to_check if col in false_negatives.columns]
if cols_exist:
    display(false_negatives[cols_exist].mean().to_frame(name='Giá trị Trung bình'))
else:
    display(false_negatives.head(3))
```
